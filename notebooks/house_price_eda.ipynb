```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
warnings.filterwarnings('ignore')

```

```python
df= pd.read_csv('../data/train.csv')
```

```python
plt.figure(figsize=(10, 12))

sns.countplot(
    data=df,
    x='City',
    order=df['City'].value_counts().index,
    palette='viridis',
    
)

plt.title('City-wise House Count')
plt.ylabel('Count')
plt.xlabel('City')

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 7))

sns.barplot(
    data=df,
    x='City',
    y='Price_INR_Lakhs',
    order=df.groupby('City')['Price_INR_Lakhs'].mean().sort_values(ascending=False).index,
    estimator='mean',
    errorbar=None,
    palette='viridis'
)



plt.tight_layout()
plt.show()
```
